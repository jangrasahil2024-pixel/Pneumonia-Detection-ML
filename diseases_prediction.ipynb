```python
normal_dir = "/content/ComputerVision/NormalXray"
pneumonia_dir= "/content/ComputerVision/PneumoniaXray"
```

```python
import os
import numpy as np
from PIL import Image

#load images
IMG_SIZE = (64,64)
X = []
y = []

# Process normal images
for file in os.listdir(normal_dir):
  img = Image.open(os.path.join(normal_dir ,file)).convert ("L")
  img = img.resize(IMG_SIZE)
  X.append(np.array(img).flatten())
  y.append(0)

# Process pneumonia images
for file in os.listdir(pneumonia_dir):
  img = Image.open(os.path.join(pneumonia_dir, file)).convert("L")
  img = img.resize(IMG_SIZE)
  X.append(np.array(img).flatten())
  y.append(1)

# Convert to numpy arrays and normalize after all data is collected
X = np.array(X) /255.0
y = np.array(y)

print(X.shape)
print(y.shape)
```

```python
# train test model
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,random_state = 2529, stratify=y )
```

```python
#build model
model = Sequential()
model.add(Dense(128,activation = "relu"))
model.add(Dense(64,activation = "relu"))
model.add(Dense(1,activation="sigmoid"))
model.compile(optimizer ="adam", loss = "binary_crossentropy",metrics= ["accuracy"])
```

```python
model.fit(X_train,y_train,epochs = 10)
```

```python
# predict
y_pred = np.round(model.predict(X_test))
```

```python
#evaluate
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
